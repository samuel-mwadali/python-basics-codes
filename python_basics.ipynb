# String
A string is a series of characters.

```python
a = "This is a string"
b = 'This is also a string'
print(a)
print(b)
```

# Changing case in a string with methods

```python
name = "Jocelyn Samuel"
print(name.title())
print(name.upper())
print(name.lower())
print(name.swapcase())
print(name.capitalize())
```

# Combining and concatenating strings
Symbol + is used to combine strings

```python
first_name = "jocelyn"
last_name = "samuel"
full_name = first_name + " " + last_name
print(full_name)
```

# Stripping whitespace
Strip white space from both sides at once use strip(), from the left side of a string lstrip() method and rtrip() method from right.

```python
name = "  samuel  "
print(name.strip())
print(name.rstrip())
print(name.lstrip())
```

# INTRODUCING LISTS
A lists is a collection of items in a particular order.<br>
You can make a list that includes letters of the alphabet, the digits from 0-9, names of people in your family.<br> Square brackets[ ] indicate a list, individual elements in the list are separated by commas.

```python
city_names =['Nairobi', 'Kisumu', 'Nakuru', 'Mombasa', 'Naivasha']
print(city_names)
```

# Acessing elements in a list
Access any element in a list by index, of the item desired.

```python
city_names = ["Nairobi", "Kisumu", "Nakuru", "Naivasha", "Mombasa"]
print(city_names[0])
print(city_names[3])
print(city_names[-1].upper()) # Acessing last element
```
